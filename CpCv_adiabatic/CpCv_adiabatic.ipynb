## 1. Heat Capacities and the Ratio γ

### 1.1 Two heat capacities
The **first law of thermodynamics** says the internal energy $U$ of a closed system changes when heat $q$ flows in or work $w$ is done on it:

$$ dU = \delta q + \delta w = \delta q - P\,dV \qquad \text{(only pressure–volume work)}$$

**Constant volume.** If $V$ is fixed, $dV = 0$, so all the heat goes into internal energy:

$$ C_V = \left(\frac{\delta q}{dT}\right)_V = \left(\frac{\partial U}{\partial T}\right)_V $$

**Constant pressure.** The enthalpy is $H = U + PV$, so $dH = \delta q + V\,dP$. At constant $P$:

$$ C_P = \left(\frac{\delta q}{dT}\right)_P = \left(\frac{\partial H}{\partial T}\right)_P $$

At constant pressure the gas expands as it warms and **does work on its surroundings**. Some of the added heat leaves as work, so more heat is needed for each degree of warming: $C_P > C_V$. For one mole of an **ideal gas**, $H = U + RT$, so

$$ C_{P,m} - C_{V,m} = R \qquad\text{and}\qquad \gamma \equiv \frac{C_{P,m}}{C_{V,m}} = 1 + \frac{R}{C_{V,m}} $$

Measuring $\gamma$ therefore tells us $C_{V,m}$. That tells us **how many ways the molecules can store energy**.

### 1.2 The equipartition theorem
Classical statistical mechanics gives every *quadratic* term in a molecule's energy an average of $\tfrac12 k_BT$, or $\tfrac12 RT$ per mole. Each term adds $\tfrac12 R$ to $C_{V,m}$:

| Motion | Quadratic terms | Contribution to $C_{V,m}$ |
|---|---|---|
| Translation ($x, y, z$) | 3 kinetic | $\tfrac32 R$ |
| Rotation, linear molecule | 2 kinetic | $R$ |
| Rotation, nonlinear molecule | 3 kinetic | $\tfrac32 R$ |
| Each vibrational mode | 1 kinetic + 1 potential | $R$ |

A molecule with $N$ atoms has $3N-5$ vibrational modes if it is linear and $3N-6$ if it is nonlinear. ([Atkins, *Physical Chemistry*, 11th ed., §2A and §13E](https://global.oup.com/academic/product/atkins-physical-chemistry-9780198769866); [McQuarrie & Simon, *Physical Chemistry: A Molecular Approach*, Ch. 17–18](https://uscibooks.aip.org/books/physical-chemistry-a-molecular-approach/))

> **Example.** Argon (monatomic) has translation only, so $C_{V,m} = \tfrac32 R$ and $\gamma = 1 + R/(\tfrac32R) = 5/3$.

## 2. Beyond Equipartition: Quantized Vibrations

Equipartition assumes energy is continuous. Vibrational energy is actually **quantized**, $E_v = (v+\tfrac12)h c\tilde\nu$. If the level spacing $hc\tilde\nu$ is much larger than $k_BT$, almost every molecule sits in $v=0$ and that vibration can't take up heat: it is "frozen out". A harmonic oscillator's contribution to $C_{V,m}$ is the **Einstein function**:

$$ \frac{C_{V,m}^{\text{vib}}}{R} = \left(\frac{\Theta_v}{T}\right)^2 \frac{e^{\Theta_v/T}}{\left(e^{\Theta_v/T}-1\right)^2}, \qquad \Theta_v = \frac{h c \tilde\nu}{k_B} \approx (1.4388\ \text{cm K})\,\tilde\nu $$

where $\Theta_v$ is the **vibrational temperature**. When $T \gg \Theta_v$ the function approaches 1, the equipartition value. When $T \ll \Theta_v$ it approaches 0. ([McQuarrie & Simon, §18-3](https://uscibooks.aip.org/books/physical-chemistry-a-molecular-approach/))

| Molecule | Mode | $\tilde\nu$ (cm⁻¹) |
|---|---|---|
| N₂ | stretch | 2358.6 |
| CO₂ | bend (doubly degenerate) | 667.4, 667.4 |
| CO₂ | symmetric stretch | 1333 |
| CO₂ | asymmetric stretch | 2349.2 |

*Frequencies: [NIST Computational Chemistry Comparison and Benchmark Database (CCCBDB)](https://cccbdb.nist.gov/), experimental fundamentals; Herzberg,* Molecular Spectra and Molecular Structure *Vol. II & IV.*

## 3. The Clément–Desormes Experiment

### 3.1 Apparatus and procedure
A carboy (a large glass jug of about 20 L) sits in a water bath. A stopper with a tube runs to an **oil manometer**. The carboy is filled with the test gas at slightly **more** than atmospheric pressure.

| State | What you do | Pressure | Temperature |
|---|---|---|---|
| **1** | Let the gas reach bath temperature; read the manometer | $P_1 = P_\text{atm} + \Delta P_1$ | $T_1$ (bath) |
| **1 → 2** | Pull the stopper **quickly**, then replace it as soon as the gas stops rushing out (≈ 1 s) | $P_2 = P_\text{atm}$ | $T_2 < T_1$ |
| **2 → 3** | Wait several minutes while the cold gas warms back to bath temperature at constant volume | $P_3 = P_\text{atm} + \Delta P_3$ | $T_1$ |

The expansion is so fast that almost no heat flows, so **1 → 2 is adiabatic** and the gas cools. With the stopper back in, the gas warms at **constant volume** and its pressure rises to $P_3$. States 1 and 3 are at the same temperature.

### 3.2 Deriving the working equation
Follow one mole of the gas that **stays** in the carboy. Let $V_1$ and $V_2$ be its molar volumes in states 1 and 2.

*Reversible adiabatic step* (1 → 2), ideal gas: $P V^\gamma = \text{constant}$, so

$$ P_1 V_1^{\gamma} = P_2 V_2^{\gamma} \quad\Longrightarrow\quad \frac{P_1}{P_2} = \left(\frac{V_2}{V_1}\right)^{\gamma} $$

*States 1 and 3 have the same temperature* and $V_3 = V_2$ (constant volume), so Boyle's law gives

$$ P_1 V_1 = P_3 V_2 \quad\Longrightarrow\quad \frac{V_2}{V_1} = \frac{P_1}{P_3} $$

Substitute and take logarithms:

$$ \boxed{\;\gamma = \frac{\ln(P_1/P_2)}{\ln(P_1/P_3)}\;} $$

**Small-pressure approximation.** The manometer readings are only about 2–3% of $P_\text{atm}$, and $\ln(1+x) \approx x$ for small $x$. Since $P_2 = P_\text{atm}$:

$$ \gamma \approx \frac{P_1 - P_2}{P_1 - P_3} = \frac{\Delta P_1}{\Delta P_1 - \Delta P_3} = \frac{h_1}{h_1 - h_3} $$

where $h_1$ and $h_3$ are the manometer height differences in states 1 and 3. ([Garland, Nibler & Shoemaker, *Experiments in Physical Chemistry*, 8th ed., Exp. 3](https://www.mheducation.com/highered/product/experiments-physical-chemistry-garland-nibler/M9780072828429.html); [Tatum, *Heat and Thermodynamics*, §8.5 "The Clément–Desormes Experiment" (LibreTexts)](https://phys.libretexts.org/Bookshelves/Thermodynamics_and_Statistical_Mechanics/Heat_and_Thermodynamics_(Tatum)/08%3A_Heat_Capacity_and_the_Expansion_of_Gases/8.05%3A_The_Clement-Desormes_Experiment); N. Clément & C. B. Desormes, *J. Phys.* **89**, 321 (1819))

### 3.3 See it on a P–V diagram
**Run the cell, then drag the sliders.** The gas follows the **adiabat** (orange, steeper) from 1 to 2, then rises straight up at constant volume to point 3 on the **isotherm** (blue). Watch how $P_3$ depends on γ.

````python
# Just run this cell, then drag the sliders.
def pv_diagram(gamma, P1_mmHg, Patm_mmHg=760.0):
    V1 = 1.0                                     # molar volume in state 1 (relative units)
    V2 = V1 * (P1_mmHg / Patm_mmHg)**(1/gamma)   # adiabatic expansion to P_atm
    P3 = P1_mmHg * V1 / V2                       # back on the isotherm through state 1
    V = np.linspace(V1 - 0.15 * (V2 - V1), V2 + 0.3 * (V2 - V1), 200)
    fig, ax = plt.subplots(figsize=(8, 5))
    ax.plot(V, P1_mmHg * V1 / V, label='isotherm through 1 ($PV$ = const)')
    ax.plot(V, P1_mmHg * (V1 / V)**gamma, label=r'adiabat through 1 ($PV^\gamma$ = const)')
    ax.annotate('', xy=(V2, Patm_mmHg), xytext=(V1, P1_mmHg),
                arrowprops=dict(arrowstyle='->', color='black', lw=2))
    ax.annotate('', xy=(V2, P3), xytext=(V2, Patm_mmHg),
                arrowprops=dict(arrowstyle='->', color='black', lw=2, ls='--'))
    for (v, p, lab) in [(V1, P1_mmHg, '1'), (V2, Patm_mmHg, '2'), (V2, P3, '3')]:
        ax.scatter(v, p, s=80, color='black', zorder=4)
        ax.text(v, p + 0.03 * (P1_mmHg - Patm_mmHg), f'  {lab}', fontsize=14, fontweight='bold')
    ax.axhline(Patm_mmHg, color='gray', lw=1, ls=':')
    dP = P1_mmHg - Patm_mmHg
    ax.set_ylim(Patm_mmHg - 0.35 * dP, P1_mmHg + 0.2 * dP)
    ax.set_xlabel('Molar volume of gas that stays in the carboy (relative to state 1)')
    ax.set_ylabel('Pressure (mmHg)')
    ax.set_title(f'γ = {gamma:.2f}:   P₃ = {P3:.2f} mmHg   (ΔP₃ = {P3 - Patm_mmHg:.2f} mmHg)')
    ax.legend(loc='upper right')
    plt.show()

interact(pv_diagram,
         gamma=FloatSlider(min=1.10, max=1.70, step=0.01, value=1.40, description='γ'),
         P1_mmHg=FloatSlider(min=765, max=800, step=1, value=780, description='P₁ (mmHg)'),
         Patm_mmHg=fixed(760.0));
````

## 6. Statistics and Uncertainty

### 6.1 Is one trial an outlier? Dixon's Q-test
A single bad expansion can give a very different γ, for example if the stopper went back in too late or the gas sloshed. Dixon's **Q-test** asks whether the most extreme value is too far from its neighbor to belong to the set ([Dean & Dixon, *Anal. Chem.* **23**, 636 (1951)](https://doi.org/10.1021/ac60052a025); critical values from [Rorabacher, *Anal. Chem.* **63**, 139 (1991)](https://doi.org/10.1021/ac00002a010)):

$$ Q = \frac{\text{gap}}{\text{range}} = \frac{|x_\text{suspect} - x_\text{nearest}|}{x_\text{max} - x_\text{min}} $$

If $Q > Q_\text{crit}$ for your number of trials $n$ and chosen confidence level, you may reject the suspect value. Use the test **once** per data set, and report every rejection.

| $n$ | 3 | 4 | 5 | 6 | 7 | 8 | 9 | 10 |
|---|---|---|---|---|---|---|---|---|
| $Q_\text{crit}$ (90%) | 0.941 | 0.765 | 0.642 | 0.560 | 0.507 | 0.468 | 0.437 | 0.412 |
| $Q_\text{crit}$ (95%) | 0.970 | 0.829 | 0.710 | 0.625 | 0.568 | 0.526 | 0.493 | 0.466 |
| $Q_\text{crit}$ (99%) | 0.994 | 0.926 | 0.821 | 0.740 | 0.680 | 0.634 | 0.598 | 0.568 |

### <font color=blue> **Question 6.1.** </font>
Complete `q_statistic`. Sort the values, find the gap at the **low** end ($x_1 - x_0$) and at the **high** end ($x_{n-1} - x_{n-2}$), and return the larger gap divided by the range, along with the suspect value at that end.

🐍 **Python tip:** `np.sort(x)` returns a sorted copy. `x[0]` is the first item, `x[1]` the second, `x[-1]` the last, `x[-2]` second-to-last.

````python
Q_CRIT = {90: [0.941, 0.765, 0.642, 0.560, 0.507, 0.468, 0.437, 0.412],
          95: [0.970, 0.829, 0.710, 0.625, 0.568, 0.526, 0.493, 0.466],
          99: [0.994, 0.926, 0.821, 0.740, 0.680, 0.634, 0.598, 0.568]}   # n = 3 … 10

def q_statistic(values):
    """Return (Q, suspect value) for the most extreme point in `values`."""
    x = np.sort(values)
    data_range = ...
    gap_low  = ...
    gap_high = ...
    if gap_high >= gap_low:
        return gap_high / data_range, x[-1]
    else:
        return gap_low / data_range, x[0]

def q_test(values, confidence=95):
    """Print and return the Q-test decision for the most extreme value."""
    n = len(values)
    if n < 3 or n > 10:
        print(f"Q-test needs 3–10 values (you have {n}).")
        return None
    Q, suspect = q_statistic(values)
    Q_crit = Q_CRIT[confidence][n - 3]
    reject = Q > Q_crit
    print(f"n = {n}: suspect γ = {suspect:.4f},  Q = {Q:.3f},  Q_crit({confidence}%) = {Q_crit:.3f}  "
          f"→ {'REJECT' if reject else 'keep'}")
    return reject, suspect

q_test(make_array(1.40, 1.41, 1.42, 1.44, 1.53, 1.76))
````

````python
def q_test_multiple(values, confidence=95):
    """
    Sequential Dixon Q-test.
    Returns (clean_values, rejected_values).
    """
    vals = np.sort(np.array(values, dtype=float))
    rejected = []

    while True:
        n = len(vals)
        if n < 3 or n > 10:
            # Q-test table only defined for n = 3–10
            break

        # Compute Q and suspect value (using same logic as your q_statistic function)
        data_range = vals[-1] - vals[0]
        gap_low  = vals[1] - vals[0]
        gap_high = vals[-1] - vals[-2]

        if gap_high >= gap_low:
            Q = gap_high / data_range
            suspect = vals[-1]
        else:
            Q = gap_low / data_range
            suspect = vals[0]

        Qcrit = Q_CRIT[confidence][n - 3]

        if Q > Qcrit:
            # Reject and continue
            rejected.append(suspect)
            vals = vals[vals != suspect]
        else:
            # No more outliers
            break

    return vals, rejected
````

### 6.2 Mean and 95% confidence interval

$$ \bar\gamma \pm t_{0.975,\,n-1}\,\frac{s}{\sqrt n} $$
